```python
df_afluencia["id_afluencias"] = df_afluencia["fecha"] + "_" + df_afluencia["hora"].astype(str) + "_" + df_afluencia["id_biblioteca"]
df_afluencia
```

```python
df_afluencia["id_afluencias"].duplicated().sum()
```

Las llaves propuestas en bibilioteca y afluencias no presentan ningun duplicado, esto es importante para que el merge no de error, porque si existe alguna llave duplicada merge simplemente multiplica y no muestra el error dandonos como resultado una tabla inservible.

```python
df_prestamos["id_afluencias"] = df_prestamos["fecha"] + "_" + df_prestamos["hora"].astype(str) + "_" + df_prestamos["id_biblioteca"]
df_prestamos
```

```python

merge_PB_left = df_prestamos.merge(df_bibliotecas, on= "id_biblioteca", how="left")
merge_PB_inner = df_prestamos.merge(df_bibliotecas, on= "id_biblioteca", how="inner")
merge_PB_outer = df_prestamos.merge(df_bibliotecas, on= "id_biblioteca", how="outer")
merge_PB_left
```

```python
merge_PB_left["id_prestamo"].duplicated().sum()
```
